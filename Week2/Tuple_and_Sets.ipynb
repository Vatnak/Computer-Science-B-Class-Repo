# Tuple

1. Tuple is a built-in data types in Python used to store collections of data.
2. Tuples are similar to lists. 
3. This collection also has iterable, ordered, and (can contain) repetitive data, just like lists.
But unlike lists, tuples are immutable(unchangeable).


# tuple method 
* Python has two built-in methods that you can use on tuples index() and count().


```python
#tuple method 

t1 = (1,4,3,2,4,6,2,3,5,7)
# .index()  returns the position of the first match only
print("Index of 2 is:", t1.index(2)) # number 2 appear in index number 3

# .count() return how many time the value appear.
print("the frequence of number 2 is : ", t1.count(2)) # number 2 appear two time in the tuple.

#convert tuple to list in order to assign item
t2 = (2,3,5,"Hello", 4,1,3.23,21)
t2_list = list(t2) #convert tuple to list
print(t2_list)
t2_list[1] = 50
print(t2_list)
t2_tuple = tuple(t2_list)
print (t2_tuple) #convert back to tuple
```

# Sets

* Set is a built-in data types in Python used to store collections of data.
* Sets do not allow duplicates and do not maintain any specific order


```python
#Creat Set
# You can creat a set of element by enclosing the elements inside a pair of curly ({})

my_set = {"apple", "banana", "cherry"}
print(my_set)
```

# Access Set

* You cannot access items in a set by referring to an index or a key.
* But you can loop through the set items using a for loop or ask if a specified value is present in a set, by using the in keyword. 



```python
# access set

# for loop
for name in my_set:
    print(name)

# in keyword   
print("cherry" in my_set) # "in" is for checking membership like "is the value exist" in the sets
```

```python
# add item to set
print("Before sets: ", my_set)
my_set.add("chocolate")
print("after sets: ", my_set)
```

# Concatenation Set

here are several ways to join two or more sets in Python.
* The union() and update() methods joins all items from both sets.
* The intersection() method keeps ONLY the duplicates.
* The difference() method keeps the items from the first set that are not in the other set(s).
* The symmetric_difference() method keeps all items EXCEPT the duplicates.


```python
#union()
set1 = {"Messi", "Ronaldo", "Neymar"}
set2 = {1,2,3}
set3 = set1.union(set2)
print("The combination of set1 and set2: ", set3) #the order can be different.
```

```python
# Intersection()

set1 = {"Apple","Banana", "cheery"}
set2 = {"google", "microsoft", "Apple", "banana"}

set3 = set1.intersection(set2)
print("The intersection of set1 and set2 is: ", set3) #Return the duplicate in the 2 sets 
```
