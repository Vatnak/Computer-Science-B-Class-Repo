## Dictionaries 
* Used to store key/value pairs.
* Provides quick access to elements using keys.
* A dictionary cannot contain duplicate keys
* Example: my_dict = {"name": "Alice", "age": 25} g data

# Creating Dictionaries

* A dictionary is created by enclosing Key-Value Pairs within curly braces {}. Each item consists of:
    - A key followed by a colon:
    - A corresponding value

* Item are seperated by commas , 
* Example :

```python
students = {"111-34-3434": "John", "132-56-6290": "Peter"}
```

* Key: 111-34-3434
* Value: "John"

# Note 
* Curly braces {} are used for both dictionaries and sets.
* {} denotes an empty dictionary.
* To create an empty set, use set().

# Key Points 

* Hashable Keys:
    - Key most be hashable (e.g., numbers, strings).
    - Values can be of any type.
    

# 1. Adding
 
* You can add a new key-value pair to a dictionary using the sysntax:
    dictionaryName[key] = value
* Example:

```python
students = {"111-34-3434": "John", "132-56-6290": "Peter"}
students["234-56-9010"] = "Susan" # added new key-value pair 
print(list(students.values())) #print only the value
print(list(students.keys())) #print only the key
print(students) #print everything 
print(students["132-56-6290"]) #access item using key
```

```python
students["111-34-3434"] = "Nak" #modify existing item
print(students["111-34-3434"])
print(students)
```

```python
print(f"Before dictionaries : {students}")
del students["234-56-9010"] #delete key-value pair
print(f"after dictionaries :{students}")
```

```python
#safer way to delete item

if "343-45-5455" in students:
    del students["343-45-5455"]
else:
    print("Key not found!")
```

# Dictionary Methods

1. Key()
    * Return: a sequence of the dictionary's keys as a tuple.
2. values()
    * Returns: a sequence of the dictionary's value as a tuple
3. items()
    * Returns: A sequence of tuples, where each tuple is in the form of (key: value)
4. get(key)
    * Returns: The value for the specified key.
    * if the key does not exist, it return None instead of raising a KeyError
5. pop(key)
    * Returns: the value for the specified key, and remove the item from the dictionary
6. popitem()
    * Returns: A randomly selected (key, value) pair as a tuple, and removes the selected item from the dictionary.
7. clear()
    * Returns: Clear the dictionary entirely
