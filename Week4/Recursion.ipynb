# Iteration vs Recursion — Simple Notes

## Iteration
- Repeating a task using a **loop** (`for`/`while`).
- Keeps track of progress using variables that update each pass.
- Runs in **one** function call — no calls pile up.
- Stops when the **loop condition** becomes false.
- Uses **less memory**.
- Generally **faster** (no extra function-call overhead).
- Risk: **infinite loop** if the condition never becomes false.
- Best for: simple repeated tasks (counting, summing, going through a list).

## Recursion
- A function that **calls itself**, working on a smaller version of the problem each time.
- Has two parts:
  - **Base case** — simplest version of the problem, answered directly (stops the recursion).
  - **Recursive case** — breaks the problem down and calls the function again.
- Works in **two phases**:
  - **Going down** — keeps calling itself, getting closer to the base case.
  - **Coming back up** — once the base case is hit, results get combined as calls return.
- Stops when the **base case** is reached.
- Uses **more memory** — each call stays on the call stack until it returns.
- Generally **slower** (call overhead).
- Risk: **stack overflow** if the base case is never reached.
- Best for: problems that are naturally self-similar (trees, nested structures, divide-and-conquer).

## Quick Comparison

| | Iteration | Recursion |
|---|---|---|
| Repeats using | Loop | Function calling itself |
| Stops at | Loop condition | Base case |
| Memory | Low | Higher (call stack) |
| Speed | Faster | Slightly slower |
| Risk | Infinite loop | Stack overflow |
| Best for | Simple repeated tasks | Self-similar problems |

**Note:** Any recursive function can be rewritten as iterative (and vice versa) — same result, different approach.

```python
def factorial_iteration(n): #Iteration function
    if n == 1 or n == 0:
        return 1
    else:
        x = 1
        for i in range(2,n+1):
            x*= i
        return x


def factorial_recusive(n): #recursive function
    if n == 0:
        return 1
    else: 
        return n * factorial_recusive(n-1)
```

```python
test1 = factorial_iteration(4)
print(f"factorial_iteration of 4 is {test1}")

test2 = factorial_recusive(4)
print(f"factorial_recursive of 4 is {test2}")
```

 **Key: Recursion can often be used when/with**:

* if-else statements that distinguish different cases.
* There exists at least one condition (base case) to stop the recursion.
* Every recursive call reduces the original problem closer to base case.
* The problems can often be written as recurrence relation

```python
#challenge

def recursive_Sn(n):
    if n == 1:
        return 1
    else: 
        return ((-1)**(n-1)/ n) + recursive_Sn(n-1)
```

```python
#Finonacci Numbers

def finoacci_numbers(n):
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        return finoacci_numbers(n-1) + finoacci_numbers(n-2)

def finoacci_numbers_sequence(count):
    return [finoacci_numbers(i) for i in range(count)]
```

```python
test4 = finoacci_numbers_sequence(5)
print(test4)
```

# Palindrome Problem 
* A string is a **palindrome** if it reads the same from the left and from the right.
* Ex: "Alexela" is plalindrome but "Alexi" is not
* **challenge**: write a recursive function **isPlaindrome(s) to check if a string s is palindrome or not


```python
def isPalindrome(s):
    if len(s) <= 1:
        return True
    elif s[0] != s[len(s) - 1]:
        return False
    else:
        return isPalindrome(s[1: len(s) - 1])
```

```python
test4 = isPalindrome("racecar")
print(test4)
```

* **isPalindrome** above is not efficient as it 
creates new string at every call.
* To avoid this, we can fix the string s and use 
another helper function called **RecursiveHelper Function.**

```python
def isPlaindrom(s):
    return isPlaindromeHelper(s, 0, len(s)-1)

def isPlaindromeHelper(s, low, high):
    if high <= low:
        return True
    elif s[low] != s[high]:
        return False
    else:
        return isPlaindromeHelper(s, low + 1, high -1)  
```

# Selection Sort

* Selection sorting is based on two ideas:
    • Find the smallest element of a list then swap it with the 
    first element of the list.
    • Ignore the first element and continue sorting the shorter 
    list.
* Ex: sort L = [3,2,5,1,2,6,3]
    • Find the smallest element = 1 then swap with 3, then L = 
    [1,2,5,3,2,6,3].
    • Do the same with smaller list L1 = [2,5,3,2,6,3]. 
    • Repeat the same process: L2 = [5,3,2,6,3] -> [2,3,5,6,3].
    • L3 = [3,5,6,3] -> L4 = [5,6,3] -> [3,6,5] -> L5 = [6,5] -> [5,6].

```python
def sort(lst):
    sortHelper(lst, 0, len(lst) - 1)

def sortHelper(lst, low, high):
    if low < high:
        indexOfMin = low
        minVal = lst[low]
        for i in range(low + 1, high + 1):
            if lst[i] < minVal:
                minVal = lst[i]
                indexOfMin = i

        lst[indexOfMin] = lst[low]
        lst[low] = minVal

        sortHelper(lst, low + 1, high)
```
