```python
sentence = input("Enter a sentence: ").lower()

punctuation = "'.,!?;:()[]{}\"-"
clean_sentence = sentence
for char in punctuation:
    clean_sentence = clean_sentence.replace(char, "")

words = clean_sentence.split()
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

print("Word Count: ")
for id, value in freq.items():
    print(f"{id} = {value} Time")
```

# challenging Exercise 
1.  Problem Description: 
You are tasked with creating a system to manage student information for a 
course. The system will handle: 
1.   Adding students and their grades. 
2.  Calculating the average grade of the class. 
3.  Displaying the student with the highest grade. 
4.  Displaying the list of students who passed or failed based on their 
grades. 
2.  Requirements: 
1.  Create a dictionary called courses, where each key represents a course 
name (e.g., "Math", "Physics", "Chemistry") and the value is another 
dictionary that stores student names as keys and their grades as values. 
 
2.  Each course should have 3 students, and the grades should be random 
values between 0 and 100. 
 
 
3.  Implement a function of average_grade that calculates and returns the 
average grade of all students in a course. 
 
4.  Implement a function of highest_grade that returns the name of the 
student with the highest grade in a course.
Implement a passing_students function that returns a set of names of 
students who passed (grades >= 50) in a given course. 
 
6.  Implement a function failed_students that returns a set of names of 
students who failed (grades < 50) in a given course. 


```python
dictionary = {
    "Math": {
        "Alice": 92,
        "Bob": 85,
        "Charlie": 78
    },
    "Physics": {
        "Nak": 88,
        "Messi": 91,
        "Sal": 74
    },
    "Chemistry": {
        "Panha": 95,
        "Ronaldo": 30,
        "Vid": 40
    }
}
```

```python
def average_grade(target_dict, course):
    for course, students in target_dict.items():
        return f"the average grade  in {course}: {sum(students.values()) / len(students.keys()):.2f}"

def highest_score(target_dict, course):
    for course, student in target_dict.items():
        top_student = max(student, key=student.get)
        return f"{top_student} has the highest grade in {course}"

def passing_students(target_dict, course):
    pass_students = []
    for student, grade in target_dict[course].items():
        if grade >= 50:
            pass_students.append(student)
    return f"Students passing in {course} are : {set(pass_students)} "

def failing_students(target_dict, course):
    fail_students = []
    for student, grade in target_dict[course].items():
        if grade < 50:
            fail_students.append(student)
    if not fail_students:
        return  f"Students failing in {course} are : None "
    return f"Students failing in {course} are : {set(fail_students)} "
    
```

```python
print(average_grade(dictionary, "Math"))
print(highest_score(dictionary, "Physics"))
print(passing_students(dictionary, "Physics"))
print(failing_students(dictionary, "Chemistry"))
```
