# Exercise 1

```python
box = ["book", ["pen", "pencil"], 
    [["toy"], "watch"]]
```

```python
def count_gifts(box):
    total = 0
    if len(box) == 0:
        return 0
    else:
        if isinstance(box[0], list):
            total += count_gifts(box[0])
        else:
            total += 1
    total += count_gifts(box[1:])
    return total
```

```python
print(count_gifts([]))
print(count_gifts(["book"]))
print(count_gifts(box))
print(count_gifts(([[], ["ball"],
            [["kite", "ball"]]])))
print(count_gifts([[["ring"]]]))
```

# Exercise 2

```python
def contains_gift(box, target):
    if len(box) == 0:
        return False
    elif box[0] == target:
        return True 
    elif isinstance(box[0], list):
        return contains_gift(box[0], target) or contains_gift(box[1:], target)
    else:
        return contains_gift(box[1:], target)
```

```python
print(contains_gift(box, "toy"))           # True 
print(contains_gift(box, "watch") )         # True 
print(contains_gift(box, "phone")  )        # False 
print(contains_gift(box, "Toy")     )       # False 
print(contains_gift([], "book")      )      # False 
print(contains_gift([[], ["pen"]], "pen"))  # True 
```

# Exercise 3

```python
def isPalindromeNum(n):
    digit = str(n)
    def check(left, right):
        if left >= right:
            return True
        if digit[left] != digit[right]:
            return False
        else:
            return check(left + 1, right - 1)
    return check(0, len(digit) - 1)
```

```python
number_group = ["0", "7", "11", "123454321", "1357531", "168861"]
for i in number_group:
    print(isPalindromeNum(i))

number_group2 = ["10", "112221"]
for i in number_group2:
    print(isPalindromeNum(i))

print(isPalindromeNum("1"))


# What stops it: check() stops when left >= right, or when two digits don't match.
# What shrinks: the gap between left and right shrinks by 2 each call.
# Why early False fails: returning False before comparing would reject valid
# palindromes without actually checking their digits.
# Extra test: isPalindromeNum(1) checks a single digit (left == right at start),
# catching a bug if the base case used > instead of >=.
```

```python
def unpack(box):
    if len(box) == 0:
        return box
    elif isinstance(box[0], list):
        head = unpack(box[0])
        tail = unpack(box[1:])
        return head + tail
    else:
        tail = unpack(box[1:])
        return [box[0]] + tail

```

```python
print(unpack(box))                   
print(unpack([["ball", "ball"]]))     
print(unpack([])) 
```

# Challenging Exercise 2


```python
def box_depth(box):
    if not box:              
        return 1
    best = 1                 
    for element in box:     
        if isinstance(element, list):
            depth = box_depth(element)   
            if depth + 1 > best:        
                best = depth + 1
    return best
```

```python
print(box_depth([]))                 
print(box_depth(["book", "pen"]))    
print(box_depth([["book"]]))         
print(box_depth(["book", [["toy"]]]))
print(box_depth([[]]))
```
