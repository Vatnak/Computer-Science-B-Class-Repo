## task 6



```python
import wikipedia
summary = wikipedia.summary("Python (programming lanaguage)")
with open("wiki_python.txt", "w") as f:
    content = f.write(summary)
    print("Saved Sucessfully!")
    f.close()


def counts(text):
    vowel = 0
    cons = 0
    digits = 0

    for char in text:
        if char.isalpha():
            if char.lower() in 'aeiou':
                vowel += 1
            else:
                cons += 1
        elif char.isdigit():
            digits += 1

    return vowel, cons, digits


v, c, d = counts(summary)

print(f"The number of vowels is {v}")
print(f"Number of consonants is {c}")
print(f"The number of digits is {d}")
```
