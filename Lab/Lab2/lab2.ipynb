```python
def make_tuple():
    t1 = ("Nak", 10000)
    return t1

def add_tuple(x1, x2):
    tuple_to_list = list(make_tuple())
    tuple_to_list.append(x1)
    tuple_to_list.append(x2)
    list_to_tuple = tuple(tuple_to_list)
    return list_to_tuple

def check_item_in_tuple(t1, x2):
    if x2 in t1:
        return x2 + " " + "Existed"
    else:
        return "not found!!"

def check_length(t1):
    return f"The length of the tuple is, {len(t1)}"


#test
my_tuple = make_tuple()
my_tuple = add_tuple("Please add me", "03132007")
print(my_tuple)
print(check_item_in_tuple(my_tuple, "03132007"))
print(check_length(my_tuple))

    
```

# Exercise 4


```python
def remove_item(t1):
    last_three = list(t1)[-3:]
    for i in last_three:
        t1.discard(i)
    return t1


def check_member(t1, item1):
    if item1 in t1:
        return f"{item1} is in set"


my_list = [1,2,3,4,5,6,7,8,9]
set_to_list = set(my_list)
print(set_to_list)
set_to_list.add(10)
set_to_list.add(20)
set_to_list.add(30)
print(set_to_list)
print(remove_item(set_to_list))
print(check_member(set_to_list, 6))
print(len(set_to_list))
```

# Chellenging Exeercise 1

```python
def search(t1): #Search Name in the set 
    item1 = input("Enter a name to search: ")
    if item1 in t1:
        return f"{item1} is in the set "
    else:
        return f"{item1} not found!"

def command(t1): 
    user = input("Do you want to add or remove a name: (add/remove): ").lower()
    if user == "add":
        add_name = input("Enter a name to add: ")
        t1.add(add_name)
        return f"Update Set: {t1}"
    elif user == "remove":
        remove_name = input("Enter a name to remove: ")
        t1.remove(remove_name)
        return f"Update Set: {t1}"

students = ("Nak", "Panha", "Sal", "Andy", "Vid", "Nak") #Create Tuple
students_sets = set(students) #Convert Tuple to set to remove duplicated 
```

```python
#test 

try:
    with open("Students.txt", "r") as f:
        loaded_name = f.read()
    print(f"The name in the file is {loaded_name}")
except FileNotFoundError:
    print("Error!!!")

with open("Students.txt", "w") as f:
    
    f.write(str(students_sets))
f.close()
print(f"Tuple: {students}")
print(f"Set after Duplicate: {str(students_sets)}")
print(search(students_sets))
print(command(students_sets))

# load and display the saved names
with open("Students.txt", "r") as f:
    saved_names = f.read()
print(f"Names loaded from file: {saved_names}")
```

# Challenging Exercise 2

```python
import random
import time
def dice_roll(guess = (1,3), n = 100):
    start = time.time()
    hits = 0
    for dice in range(n):
        dice = random.randint(1,6)
        if dice in guess:
            hits += 1
    end = time.time()
    running_time = end - start
    return f"The proportion is: {hits / n} and the running time is: {running_time}"


test = dice_roll(guess = (1,6), n = 100000)
print(test)
        

```
